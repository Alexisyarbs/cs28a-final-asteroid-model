```python
import pandas as pd

df = pd.read_csv("neo.csv")
df.info()
```

```python
# visualize hazardous vs non-hazardous astroiods
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(x="hazardous", data=df)
plt.title("Distribution of Hazardous vs Non-Hazardous Asteroids")
plt.savefig('haz_vs_nonhaz_dist.png', bbox_inches='tight')
plt.show()
```

```python
#scattler plot showing astroid size and distance from earth 
plt.figure(figsize=(8, 6))
sns.scatterplot(x='miss_distance', y='absolute_magnitude', hue='hazardous', data=df, alpha=0.5)
plt.title('Asteroid Size vs Distance (Colored by Hazard)')
plt.gca().invert_yaxis() # Invert because lower magnitude means a larger asteroid
plt.savefig('size_vs_dist.png', bbox_inches='tight')
plt.show()
```

```python
from sklearn.model_selection import train_test_split

# drop id, nae, orbiting_body, sentry_object, and hardous columns because they do not help predict harards
X = df.drop(['id', 'name', 'orbiting_body', 'sentry_object', 'hazardous'], axis=1)

# convert hazardous to integers
y = df['hazardous'].astype(int)

# split data for testing training
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

#create and train the model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# predictions on the test set
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print(f"Model Accuracy: {accuracy * 100:.2f}%")
```

```python
# feature importance
importances = model.feature_importances_
features = X.columns

importance_df = pd.DataFrame({'Feature': features, 'Importance': importances})
importance_df = importance_df.sort_values(by='Importance', ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x='Importance', y='Feature', data=importance_df, palette='viridis')
plt.title('Which Features Mattered Most to the Model?')
plt.savefig('features.png', bbox_inches='tight')
plt.show()

# Combine X and y temporarily just for the heat map
data_for_heatmap = X.copy()
data_for_heatmap['hazardous'] = y

plt.figure(figsize=(10, 8))

# Calculate the correlation between all columns
correlation_matrix = data_for_heatmap.corr()

# Create the heat map
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Heat Map of Asteroid Characteristics')
plt.savefig('heatmap.png', bbox_inches='tight')
plt.show()
```

```python
# columns model expects
expected_columns = X.columns

# 2. Create a dictionary for our fake asteroids using average safe values as a baseline
fake_data = pd.DataFrame(columns=expected_columns)
fake_data.loc[0] = X.mean()
fake_data.loc[1] = X.mean()

# Asteroid 0: Low absolute magnitude and low miss distance.
fake_data.loc[0, 'absolute_magnitude'] = 15.0 
fake_data.loc[0, 'miss_distance'] = 0.01 

# High absolute magnitude and high miss distance = very far.
fake_data.loc[1, 'absolute_magnitude'] = 30.0 
fake_data.loc[1, 'miss_distance'] = 0.20 

# 5. Make the predictions
predictions = model.predict(fake_data)


print("Hypothetical Asteroid Test")
print(f"Asteroid 0 (Huge & Close): {'HAZARDOUS' if predictions[0] == 1 else 'SAFE'}")
print(f"Asteroid 1 (Tiny & Far): {'HAZARDOUS' if predictions[1] == 1 else 'SAFE'}")
```
